## Phần 1 — Vì sao cần NumPy?

So sánh tốc độ giữa Python thuần và NumPy trên 1 triệu số.

```python
import time

n = 1_000_000
list_a = list(range(n))
arr_a = np.arange(n)

t0 = time.time()
tong1 = sum(x * x for x in list_a)      # Python thuan
t1 = time.time()
tong2 = np.sum(arr_a * arr_a)           # NumPy
t2 = time.time()

print(f"Python list: {t1 - t0:.4f} giay")
print(f"NumPy      : {t2 - t1:.4f} giay")
print(f"NumPy nhanh hon khoang {(t1 - t0) / (t2 - t1):.0f} lan")
print("Ket qua giong nhau?", tong1 == tong2)
```

```python
# Khac biet co ban giua list va ndarray
l = [1, 2, 3]
a = np.array([1, 2, 3])

print("list * 2  :", l * 2)     # lap lai list
print("array * 2 :", a * 2)     # nhan tung phan tu
print("list + list  :", l + l)  # noi list
print("array + array:", a + a)  # cong tung phan tu
```

## Phần 5 — Lọc dữ liệu bằng điều kiện

Boolean masking là cách "lọc" dữ liệu nhanh nhất và cũng là cách bạn sẽ dùng suốt khi xử lý dữ liệu AI.

```python
diem = np.array([8.5, 4.0, 7.0, 9.5, 5.5, 3.0, 6.0, 9.0])

mask = diem >= 5
print("mask        :", mask)
print("diem dat    :", diem[mask])
print("so hs dat   :", np.sum(diem >= 5))
print("ti le dat   :", np.mean(diem >= 5))
print("diem gioi   :", diem[(diem >= 8) & (diem <= 9.5)])
print("diem duoi TB:", diem[~(diem >= 5)])
```

### 12.2 Các hàm kích hoạt

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def softmax(x):
    e = np.exp(x - np.max(x))     # tru max de tranh tran so
    return e / np.sum(e)

x = np.linspace(-5, 5, 11)
print("x      :", x)
print("sigmoid:", np.round(sigmoid(x), 3))
print("relu   :", relu(x))
print("softmax:", np.round(softmax(np.array([2.0, 1.0, 0.1])), 3),
      "-> tong =", softmax(np.array([2.0, 1.0, 0.1])).sum())
```

### 12.3 Hàm mất mát và độ chính xác

```python
y_that = np.array([3.0, 5.0, 2.5, 7.0])
y_pred = np.array([2.8, 5.2, 3.0, 6.5])

mse = np.mean((y_that - y_pred) ** 2)
mae = np.mean(np.abs(y_that - y_pred))
print(f"MSE = {mse:.4f} | MAE = {mae:.4f}")

nhan_that = np.array([0, 1, 1, 0, 1, 1, 0, 0])
nhan_pred = np.array([0, 1, 0, 0, 1, 1, 1, 0])
print("Do chinh xac:", np.mean(nhan_that == nhan_pred))
```

### 12.4 Chia dữ liệu train / test

```python
X = rng.normal(size=(200, 4))
# Nhan co QUY LUAT that su: tong hai dac trung dau > 0 -> nhan 1
y = (X[:, 0] + X[:, 1] > 0).astype(int)

idx = np.arange(len(X))
rng.shuffle(idx)                 # xao tron CHI SO, khong xao tron du lieu

n_train = int(0.8 * len(X))
train_idx, test_idx = idx[:n_train], idx[n_train:]

X_train, y_train = X[train_idx], y[train_idx]
X_test, y_test = X[test_idx], y[test_idx]
print("Train:", X_train.shape, "| Test:", X_test.shape)
print("Ti le nhan 1 trong train:", y_train.mean().round(3))
```

### 12.5 Thuật toán k-NN — Machine Learning trong 5 dòng

```python
def knn_du_doan(X, y, mau_moi, k=5):
    kc = np.linalg.norm(X - mau_moi, axis=1)   # khoang cach toi MOI mau
    lang_gieng = np.argsort(kc)[:k]            # k mau gan nhat
    return np.bincount(y[lang_gieng]).argmax() # nhan pho bien nhat

mau_moi = rng.normal(size=4)
print("Du doan:", knn_du_doan(X_train, y_train, mau_moi, k=5))

# Do chinh xac tren tap test
du_doan = np.array([knn_du_doan(X_train, y_train, m, k=5) for m in X_test])
print("Do chinh xac tren test:", (du_doan == y_test).mean())
```

### 12.6 Hồi quy tuyến tính bằng gradient descent

Toàn bộ Deep Learning dựa trên đúng ý tưởng của ô code này.

```python
x = np.linspace(0, 10, 50)
y = 3 * x + 5 + rng.normal(0, 1, 50)      # du lieu that: w = 3, b = 5

w, b = 0.0, 0.0
lr = 0.01
lich_su = []

for buoc in range(1001):
    y_pred = w * x + b
    loi = y_pred - y
    loss = np.mean(loi ** 2)
    lich_su.append(loss)

    dw = 2 * np.mean(loi * x)
    db = 2 * np.mean(loi)
    w -= lr * dw
    b -= lr * db

    if buoc % 200 == 0:
        print(f"Buoc {buoc:4d} | loss = {loss:8.4f} | w = {w:.3f} | b = {b:.3f}")

print(f"\nKet qua: w = {w:.2f} (that: 3), b = {b:.2f} (that: 5)")
```

```python
# Ve do thi (bo qua neu chua cai matplotlib)
try:
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(1, 2, figsize=(11, 4))
    ax[0].scatter(x, y, s=18, label="du lieu")
    ax[0].plot(x, w * x + b, color="red", label="duong hoi quy")
    ax[0].legend(); ax[0].set_title("Hoi quy tuyen tinh")
    ax[1].plot(lich_su); ax[1].set_yscale("log")
    ax[1].set_xlabel("buoc"); ax[1].set_ylabel("loss")
    ax[1].set_title("Loss giam dan")
    plt.tight_layout(); plt.show()
except ImportError:
    print("Chua co matplotlib. Cai bang: pip install matplotlib")
```

```python
# --- Bai 4 ---
def chuan_hoa(X):
    mn = X.min(axis=0)
    mx = X.max(axis=0)
    return (X - mn) / (mx - mn)

X = rng.normal(size=(10, 3))
Xn = chuan_hoa(X)
print(Xn.min(axis=0), Xn.max(axis=0))

# --- Bai 5 ---
mon = np.array(["Toan", "Ly", "Hoa", "Sinh"])
diem = np.round(rng.uniform(4, 10, size=(6, 4)), 1)
dtb = diem.mean(axis=1)
xep_loai = np.where(dtb >= 8, "Gioi",
           np.where(dtb >= 6.5, "Kha",
           np.where(dtb >= 5, "TB", "Yeu")))
print(np.round(dtb, 2))
print(xep_loai)
print("Mon kho nhat:", mon[diem.mean(axis=0).argmin()])
```

```python
# --- Bai 6 ---
def mse(y_that, y_pred):
    return np.mean((y_that - y_pred) ** 2)

def mae(y_that, y_pred):
    return np.mean(np.abs(y_that - y_pred))

yt = np.array([1.0, 2.0, 3.0])
yp = np.array([1.1, 1.8, 3.5])
print(mse(yt, yp), mae(yt, yp))

# --- Bai 7 ---
img = rng.integers(0, 256, size=(64, 64, 3), dtype=np.uint8)
gray = img.mean(axis=2)
flip = img[:, ::-1, :]
crop = img[16:48, 16:48]
sang = np.clip(img.astype(np.int16) + 30, 0, 255).astype(np.uint8)
print(gray.shape, flip.shape, crop.shape, sang.max())
```

```python
# --- Bai 8 ---
n = 25
X0 = rng.normal([0, 0], 1, size=(n, 2))
X1 = rng.normal([4, 4], 1, size=(n, 2))
X = np.vstack([X0, X1])
y = np.concatenate([np.zeros(n, dtype=int), np.ones(n, dtype=int)])

def knn(X, y, diem_moi, k=3):
    kc = np.linalg.norm(X - diem_moi, axis=1)
    return np.bincount(y[np.argsort(kc)[:k]]).argmax()

print("Diem gan nhom 0:", knn(X, y, np.array([0.5, 0.2])))
print("Diem gan nhom 1:", knn(X, y, np.array([3.8, 4.1])))

# --- Bai 9 ---
def softmax2d(x):
    e = np.exp(x - x.max(axis=1, keepdims=True))
    return e / e.sum(axis=1, keepdims=True)

Z = rng.normal(size=(3, 4))
P = softmax2d(Z)
print(np.round(P, 3))
print("Tong moi hang:", P.sum(axis=1))
```

```python
# --- Bai 10 ---
w_that = np.array([2.0, -1.0, 0.5])
b_that = 3.0
X = rng.normal(size=(100, 3))
y = X @ w_that + b_that + rng.normal(0, 0.1, 100)

w = np.zeros(3)
b = 0.0
lr = 0.05
for buoc in range(2000):
    loi = (X @ w + b) - y
    w -= lr * 2 * (X.T @ loi) / len(y)
    b -= lr * 2 * loi.mean()

print("w hoc duoc:", np.round(w, 3), " | w that:", w_that)
print("b hoc duoc:", round(b, 3), " | b that:", b_that)
```
