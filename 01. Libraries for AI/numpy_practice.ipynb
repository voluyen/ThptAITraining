## Phần 1 — Vì sao cần NumPy?

So sánh tốc độ giữa Python thuần và NumPy trên 1 triệu số.

```python
import time

n = 1_000_000
list_a = list(range(n))
arr_a = np.arange(n)

t0 = time.time()
tong1 = sum(x * x for x in list_a)      # pure Python
t1 = time.time()
tong2 = np.sum(arr_a * arr_a)           # NumPy
t2 = time.time()

print(f"Python list: {t1 - t0:.4f} giay")
print(f"NumPy      : {t2 - t1:.4f} giay")
print(f"NumPy nhanh hon khoang {(t1 - t0) / (t2 - t1):.0f} lan")
print("Ket qua giong nhau?", tong1 == tong2)
```

```python
# Khac biet co ban giua list va ndarray
l = [1, 2, 3]
a = np.array([1, 2, 3])

print("list * 2  :", l * 2)     # repeats the list
print("array * 2 :", a * 2)     # element-wise product
print("list + list  :", l + l)  # concatenates
print("array + array:", a + a)  # element-wise sum
```

Boolean masking là cách "lọc" dữ liệu nhanh nhất và cũng là cách bạn sẽ dùng suốt khi xử lý dữ liệu AI.

### 11.2 Nhân ma trận

Phần tử ở **hàng i, cột j** của kết quả = tích vô hướng của **hàng i** (ma trận trái)
với **cột j** (ma trận phải).

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print("A * B (tung phan tu):\n", A * B)
print("A @ B (nhan ma tran):\n", A @ B)

# Check the element at row 0, column 0 by hand
print("Hang 0 cua A:", A[0], " Cot 0 cua B:", B[:, 0])
print("Tich vo huong cua chung:", np.dot(A[0], B[:, 0]), "= A @ B tai [0,0]")
```

### 11.4 Hệ phương trình dưới dạng ma trận

$\begin{cases} 2x + y = 5 \\\\ x + 3y = 10 \end{cases}$
$\Longleftrightarrow$
$\begin{pmatrix} 2 & 1 \\\\ 1 & 3 \end{pmatrix}
 \begin{pmatrix} x \\\\ y \end{pmatrix} =
 \begin{pmatrix} 5 \\\\ 10 \end{pmatrix}$

Nhân hàng 1 với vector ẩn cho ra $2x + y$ — đúng vế trái của phương trình thứ nhất.

```python
A = np.array([[2, 1],
              [1, 3]])
b = np.array([5, 10])

nghiem = np.linalg.solve(A, b)
print("x, y =", nghiem)                  # [1. 3.]
print("Check A @ solution =", A @ nghiem, " (must equal b)")
```

### 12.2 Hàm softmax

$\mathrm{softmax}(x)_i = \dfrac{e^{x_i}}{\sum_j e^{x_j}}$ — biến một dãy số bất kỳ
thành các số dương có tổng bằng 1.

```python
def softmax(x):
    e = np.exp(x - np.max(x))     # subtract max: avoid overflow
    return e / np.sum(e)

diem_so = np.array([2.0, 1.0, 0.1])
p = softmax(diem_so)
print("softmax:", np.round(p, 3))
print("tong   :", p.sum())
```

### 12.3 Đo sai số giữa hai dãy số
